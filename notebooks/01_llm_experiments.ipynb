```python
from database.database import get_quiz_results

USER_ID = 1

results = get_quiz_results(USER_ID)

print(f"Total quiz results: {len(results)}")
```

```python
import pandas as pd

results_data = [dict(row) for row in results]

quiz_df = pd.DataFrame(results_data)

quiz_df
```

```python
if quiz_df.empty:
    print("No quiz data available yet.")
else:
    print(
        "Average Score:",
        round(quiz_df["percentage"].mean(), 2),
        "%"
    )

    print(
        "Best Score:",
        round(quiz_df["percentage"].max(), 2),
        "%"
    )

    print(
        "Lowest Score:",
        round(quiz_df["percentage"].min(), 2),
        "%"
    )
```

```python
import matplotlib.pyplot as plt

if not quiz_df.empty:

    plt.figure(figsize=(10, 5))

    plt.bar(
        quiz_df["topic"],
        quiz_df["percentage"]
    )

    plt.xlabel("Quiz Topic")
    plt.ylabel("Score (%)")
    plt.title("Quiz Performance")

    plt.xticks(rotation=30)
    plt.tight_layout()

    plt.show()
```
